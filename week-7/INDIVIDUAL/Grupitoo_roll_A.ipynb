```python
import pandas as pd

# --- SAMM 1 & 2: Impordi pandas ja loe CSV failid ---
df_sales = pd.read_csv('datasets/clean/sales.csv')
df_customers = pd.read_csv('datasets/clean/customers.csv')

# --- SAMM 3: Sales tabel (shape + head) ---
print("=== SAMM 3: Sales tabel ===")
print("Sales shape:", df_sales.shape)
display(df_sales.head())

# --- SAMM 4: Customers tabel (shape + head) ---
print("\n=== SAMM 4: Customers tabel ===")
print("Customers shape:", df_customers.shape)
display(df_customers.head())

# --- SAMM 5: Tabelite liitmine (Left Join customer_id põhjal) ---
df = pd.merge(df_sales, df_customers, on='customer_id', how='left')

# --- SAMM 6: Liidetud DataFrame'i kontroll (shape, dtypes, head ja veerud) ---
print("\n=== SAMM 6: Liidetud DataFrame (df) ===")
print("Liidetud tabeli suurus (shape):", df.shape)

print("\n1. Andmetüübid (dtypes):")
print(df.dtypes)

print("\n2. Kohustuslike veergude olemasolu kontroll:")
req_cols = ['customer_id', 'sale_date', 'total_price', 'email']
for col in req_cols:
    olemas = col in df.columns
    print(f"  - Veerg '{col}': {'✅ OLEMAS' if olemas else '❌ PUUDUB'}")

print("\n3. Liidetud tabeli esimesed ridad (head):")
display(df.head())

# Salvestame lõpliku tulemuse Roll B jaoks
df.to_csv('df_merged_roll_A_loplik.csv', index=False)
print("\n✅ Fail 'df_merged_roll_A_loplik.csv' on edukalt loodud ja salvestatud!")
```
